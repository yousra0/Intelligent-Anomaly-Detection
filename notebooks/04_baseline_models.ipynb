### 2. Chargement des données préparées

Chargement depuis `data/processed/` (produit par NB02).
Les données sont déjà scalées et prêtes — aucune transformation supplémentaire nécessaire.

```python
# — Chargement features.json et class_weights.json —
with open(MODELS_DIR / 'features.json', encoding='utf-8') as f:
    meta = json.load(f)
with open(MODELS_DIR / 'class_weights.json', encoding='utf-8') as f:
    class_weights_raw = json.load(f)

FEATURE_COLS  = meta['all_features']
TARGET        = meta['target']
CLASS_WEIGHTS = {int(k): float(v) for k, v in class_weights_raw.items()}

print(f'Features ({len(FEATURE_COLS)}) : {FEATURE_COLS}')
print(f'Target              : {TARGET}')
print(f'class_weights       : {CLASS_WEIGHTS}')
```

```python
# — Chargement des splits (NPY → DataFrame) —
# Compatibilité : certains artefacts .npy historiques sont sauvegardés en dtype=object.
# On force le cast vers float32/int64 avant de créer les DataFrames.
def load_split(prefix):
    X_arr = np.load(PROCESSED_DIR / f'X_{prefix}.npy', allow_pickle=True)
    y_arr = np.load(PROCESSED_DIR / f'y_{prefix}.npy', allow_pickle=True)

    if X_arr.dtype == object:
        X_arr = X_arr.astype(np.float32)
    if y_arr.dtype == object:
        y_arr = y_arr.astype(np.int64)

    # Validation : s'assure que le split est cohérent avec features.json
    if X_arr.shape[1] != len(FEATURE_COLS):
        raise ValueError(
            f"Mismatch features: X_{prefix}.npy a {X_arr.shape[1]} colonnes, "
            f"features.json en définit {len(FEATURE_COLS)}"
        )

    X = pd.DataFrame(X_arr, columns=FEATURE_COLS)
    y = pd.Series(np.asarray(y_arr).ravel().astype(int), name=TARGET)
    return X, y

X_train, y_train = load_split('train')
X_val,   y_val   = load_split('val')
X_test,  y_test  = load_split('test')
X_smote, y_smote = load_split('train_smote')  # ratio ~1:10 après SMOTE

print('Datasets chargés :')
print(f'  Train         : {X_train.shape}  fraudes={int(y_train.sum())}  ({y_train.mean()*100:.4f}%)')
print(f'  Val           : {X_val.shape}    fraudes={int(y_val.sum())}')
print(f'  Test          : {X_test.shape}   fraudes={int(y_test.sum())}')
print(f'  Train SMOTE   : {X_smote.shape}  fraudes={int(y_smote.sum())}  ({y_smote.mean()*100:.4f}%)')
print('\nDonnées chargées avec succès.')
```

```python
X_normal = np.load(PROCESSED_DIR / 'X_train_normal.npy', allow_pickle=True)
y_normal = np.load(PROCESSED_DIR / 'y_train_normal.npy', allow_pickle=True)

if X_normal.dtype == object:
    X_normal = X_normal.astype(np.float32)
if y_normal.dtype == object:
    y_normal = y_normal.astype(np.int64)

print(f' Train normal : {X_normal.shape}  fraudes={int(y_normal.sum())}  (AutoEncoder)')
```

### 3. Logistic Regression — `LR_balanced`

**Stratégie :** `class_weight='balanced'` pénalise les erreurs sur les fraudes
d'un facteur **386.74×** sans modifier les données d'entraînement.

Évite l'overfitting sur une classe positive très peu représentée.

### 7. XGBoost — `XGB_smote`

**Stratégie :** Entraîné sur `X_smote` (données SMOTE). 
`scale_pos_weight=1` car SMOTE équilibre déjà les classes.  
Généralement le meilleur modèle supervisé sur données tabulaires.

> Pas de wrapper custom — utilisation directe de `XGBClassifier` sklearn-compatible.

```python
from xgboost import XGBClassifier

t0 = time.time()
xgb_smote = XGBClassifier(
    n_estimators     = 300,
    max_depth        = 6,
    learning_rate    = 0.1,
    subsample        = 0.8,
    colsample_bytree = 0.8,
    scale_pos_weight = 1,          # SMOTE equilibre deja les classes
    random_state     = RANDOM_STATE,
    eval_metric      = 'aucpr', #plus pertinent que logloss
    n_jobs           = -1,
)
xgb_smote.fit(
    X_smote, y_smote,              # memes noms que dans NB03
    eval_set = [(X_val, y_val)],
    verbose  = False,
)
xgb_train_time = round(time.time() - t0, 2)

print(f'XGBClassifier')
print(f'  n_estimators={xgb_smote.n_estimators}')
print(f'  max_depth={xgb_smote.max_depth}')
print(f'  learning_rate={xgb_smote.learning_rate}')
print(f'  train_time={xgb_train_time}s')
print(f'  n_features_in={xgb_smote.n_features_in_}')
```

### 8. Isolation Forest — `IsoForest`

**Stratégie :** Modèle **non-supervisé**. `contamination` = taux de fraude réel (~0.13%).  
Entraîné sur **toutes** les transactions du train (labels non utilisés).

> **Intérêt clé :** comparaison directe avec l'AutoEncoder (NB04) —  
> ces deux modèles fonctionnent sans labels, argument fort pour la production.

```python
from sklearn.ensemble      import IsolationForest
from sklearn.preprocessing import MinMaxScaler

# contamination = taux de fraude réel dans le train (~0.13%)
# Ce paramètre indique à l'IF le pourcentage attendu d'anomalies.
contamination = round(float(y_train.mean()), 6)
print(f'contamination = {contamination}  ({contamination*100:.4f}%)')

t0 = time.time()
iso_forest = IsolationForest(
    n_estimators  = 200,
    contamination = contamination,
    random_state  = RANDOM_STATE,
    n_jobs        = -1,
)
# Entraînement sans labels : l'IF apprend la structure normale des données
iso_forest.fit(X_train)
iso_train_time = round(time.time() - t0, 2)
print(f'IsolationForest  n_estimators={iso_forest.n_estimators}  train_time={iso_train_time}s')

# score_samples() retourne des valeurs négatives (plus négatif = plus anormal).
# On inverse le signe pour que scores élevés = transactions suspectes.
# MinMaxScaler normalisé sur X_train uniquement pour éviter le data leakage.
# Note : les scores de val/test peuvent dépasser 1.0 si une transaction
# est plus anormale que tout ce que le modèle a vu en train — comportement attendu.
raw_train = -iso_forest.score_samples(X_train)
raw_val   = -iso_forest.score_samples(X_val)
raw_test  = -iso_forest.score_samples(X_test)

scaler_if = MinMaxScaler()
scaler_if.fit(raw_train.reshape(-1, 1))  # fit sur train uniquement

iso_score_val  = scaler_if.transform(raw_val.reshape(-1, 1)).ravel()
iso_score_test = scaler_if.transform(raw_test.reshape(-1, 1)).ravel()

print(f'Scores val  : min={iso_score_val.min():.4f}  max={iso_score_val.max():.4f}')
print(f'Scores test : min={iso_score_test.min():.4f}  max={iso_score_test.max():.4f}')

iso_pred_val  = (iso_score_val  >= 0.5).astype(int)
iso_pred_test = (iso_score_test >= 0.5).astype(int)

metrics_if_val  = compute_fraud_metrics(
    np.asarray(y_val),  iso_pred_val,  iso_score_val,
    threshold=0.5, model_name='IsoForest (val)'
)
metrics_if_test = compute_fraud_metrics(
    np.asarray(y_test), iso_pred_test, iso_score_test,
    threshold=0.5, model_name='IsoForest (test)'
)

print()
print('--- VALIDATION ---')
print_metrics_report(metrics_if_val,  BASELINE_RECALL, BASELINE_F1)
print()
print('--- TEST (seuil=0.5) ---')
print_metrics_report(metrics_if_test, BASELINE_RECALL, BASELINE_F1)
```

### 9. Analyse du seuil de décision

Le seuil par défaut (0.5) n'est pas optimal pour un dataset aussi déséquilibré.
On cherche le seuil qui **maximise le F-beta (β=2)** sur la validation, puis on
applique ce seuil au test.

> **Règle** : le seuil est toujours sélectionné sur `X_val`,
> jamais sur `X_test` pour éviter le data snooping.

```python
# — Seuils optimaux sur Val —
# On utilise F-beta avec beta=2 : le Recall compte 2× plus que la Precision.
# Justification : dans la fraude financière, manquer une fraude (FN) est
# beaucoup plus coûteux que signaler une transaction légitime (FP).
# Le seuil est TOUJOURS cherché sur X_val, jamais sur X_test.
models_val_scores = {
    'LR_balanced': (np.asarray(y_val), lr_bal_score_val),
    'LR_smote':    (np.asarray(y_val), lr_smote_score_val),
    'RF_balanced': (np.asarray(y_val), rf_bal_score_val),
    'RF_smote':    (np.asarray(y_val), rf_smote_score_val),
    'XGB_smote':   (np.asarray(y_val), xgb_score_val),
    'IsoForest':   (np.asarray(y_val), iso_score_val),
}

optimal_thresholds = {}
print(f'  {"Modèle":<16} {"Seuil opt.":>12}  {"Fbeta2 (val)":>12}')
print('  ' + '-'*44)
for name, (yt, ys) in models_val_scores.items():
    t_opt, score_opt = find_optimal_threshold(yt, ys, metric='fbeta', beta=2.0)
    optimal_thresholds[name] = t_opt
    print(f'  {name:<16} {t_opt:>12.4f}  {score_opt:>12.4f}')
```

```python
# — Réalité sur Test avec seuils optimaux —
models_test_scores_dict = {
    'LR_balanced': (np.asarray(y_test), lr_bal_score_test),
    'LR_smote':    (np.asarray(y_test), lr_smote_score_test),
    'RF_balanced': (np.asarray(y_test), rf_bal_score_test),
    'RF_smote':    (np.asarray(y_test), rf_smote_score_test),
    'XGB_smote':   (np.asarray(y_test), xgb_score_test),
    'IsoForest':   (np.asarray(y_test), iso_score_test),
}

metrics_opt = {}
for name, (yt, ys) in models_test_scores_dict.items():
    t = optimal_thresholds[name]
    y_pred_opt = (ys >= t).astype(int)
    m = compute_fraud_metrics(yt, y_pred_opt, ys, threshold=t,
                               model_name=f'{name} (t={t:.2f})')
    metrics_opt[name] = m
    print_metrics_report(m, BASELINE_RECALL, BASELINE_F1)
    print()
```

```python
# — Sauvegarde des modèles —
# Les wrappers custom (LR, RF) utilisent leur propre méthode .save() (joblib + métadonnées).
# XGBoost et IsolationForest n'ont pas de wrapper — sauvegarde joblib directe.

def save_model_with_check(model, path):
    """Sauvegarde un modèle wrapper et affiche la taille du fichier."""
    try:
        model.save(path)
        p = Path(path)
        if p.exists() and p.is_file():
            print(f'  OK  {p.name:<30} ({p.stat().st_size / 1024:.0f} Ko)')
        else:
            print(f'  WARN  {p.name}  (fichier non créé)')
    except Exception as e:
        print(f'  ERROR  {Path(path).name}: {e}')

def save_with_joblib(obj, path):
    """Sauvegarde un objet quelconque via joblib et affiche la taille du fichier."""
    try:
        joblib.dump(obj, path)
        p = Path(path)
        if p.exists() and p.is_file():
            print(f'  OK  {p.name:<30} ({p.stat().st_size / 1024:.0f} Ko)')
        else:
            print(f'  WARN  {p.name}  (fichier non créé)')
    except Exception as e:
        print(f'  ERROR  {Path(path).name}: {e}')

print('Modèles sauvegardés :')

# Wrappers custom — LR et RF
for name, model in [('lr_balanced', lr_balanced), ('lr_smote', lr_smote),
                    ('rf_balanced', rf_balanced), ('rf_smote', rf_smote)]:
    save_model_with_check(model, MODELS_DIR / f'{name}.pkl')

# Joblib direct — XGBoost, IsolationForest, scaler de normalisation
for obj, name in [(xgb_smote,  'xgb_smote'),
                  (iso_forest, 'iso_forest'),
                  (scaler_if,  'iso_forest_scaler')]:
    save_with_joblib(obj, MODELS_DIR / f'{name}.pkl')
```

## Prediction sur dataset arbitraire (schema-agnostic)

Cette section montre comment utiliser les modeles entraines pour analyser **n'importe quel CSV**
sans connaitre son schema a l'avance. Le systeme detecte automatiquement les colonnes pertinentes
et choisit le pipeline de prediction adapte (XGBoost+AE, AE seul, ou IsolationForest on-the-fly).


```python
# Imports des services backend
import sys
project_root_svc = Path(".").resolve().parent
if str(project_root_svc) not in sys.path:
    sys.path.insert(0, str(project_root_svc))

from app.services.column_mapper import map_columns
from app.services.schema_detector import detect_schema_mode
from app.services.feature_builder import build_features_dynamic
from app.services.dataset_profiler import profile_dataset
from app.services.generic_predictor import predict_generic_batch
from app.services.predictor import predict_batch

print("Services charges.")
```

```python
def analyser_csv(csv_path, models_dict):
    """
    Analyse n'importe quel CSV et retourne les transactions triees par risque.

    Parametres
    ----------
    csv_path    : chemin vers le fichier CSV
    models_dict : dict des modeles charges via load_all_models()

    Retourne
    --------
    pd.DataFrame avec colonnes : tx_id, risk_level, is_fraud_predicted,
                                  ae_score?, isoforest_score?, xgb_score?
    """
    df = pd.read_csv(csv_path)
    print(f"Fichier : {csv_path}")
    print(f"Shape   : {df.shape}")

    # Profilage
    try:
        profile = profile_dataset(df)
    except Exception:
        profile = None

    # Mapping + detection du mode
    mapping = map_columns(df)
    schema  = detect_schema_mode(mapping, df)
    df_mapped = mapping.df

    print(f"Mode    : {schema.mode}  |  Modele : {schema.model_label}")
    print(f"Raison  : {schema.reason}")

    # Construction features AE si applicable
    X_arr = None
    if schema.use_ae or schema.use_xgb:
        try:
            X_arr, _ = build_features_dynamic(df_mapped, profile, models_dict["scaler"])
        except Exception as e:
            print(f"Features AE non disponibles : {e}")
            schema.use_ae = False

    # Prediction
    if schema.mode == "standard":
        transactions = predict_batch(X_arr, models_dict, df_mapped)
    else:
        transactions = predict_generic_batch(
            df_original=df,
            X_arr=X_arr,
            models=models_dict,
            schema_result=schema,
            df_mapped=df_mapped,
        )

    df_out = pd.DataFrame(transactions)
    n_fraud = df_out["is_fraud_predicted"].sum()
    print(f"Anomalies : {n_fraud}/{len(df_out)} ({n_fraud/len(df_out)*100:.2f}%)")
    return df_out


# ── Utilisation ────────────────────────────────────────────────────────────
# Charger les modeles (une seule fois par session)
from app.services.predictor import load_all_models
models_loaded = load_all_models()

# Analyser le dataset courant (ou remplacer par n'importe quel CSV)
CSV_A_ANALYSER = PROCESSED_DIR / "X_test.csv"  # ou Path("mon_fichier.csv")
try:
    df_anomalies = analyser_csv(CSV_A_ANALYSER, models_loaded)
    df_anomalies.head(10)
except FileNotFoundError:
    print(f"Fichier non trouve : {CSV_A_ANALYSER}")
    print("Remplacez CSV_A_ANALYSER par le chemin de votre fichier CSV.")
```
